```python
# Autograd is a core component of Pytorch that provides automatic differentiations for tensor operations. 
# It enables gradient computation which is essential for training machine learning models using optimisation algos like gradient descent
```

```python
import torch
```

```python
x= torch.tensor(3.0, requires_grad=True) # Need to calculate derivative
```

```python
y=x**2
```

```python
x
y # GRAD_fn
# Pytorch makes a computation graph internally. 
```

```python
y.backward()
x.grad
```

```python
# Neural Network
x=torch.tensor(6.7)
y=torch.tensor(0.0)
```

```python
w= torch.tensor(1.0,requires_grad=True)
b= torch.tensor(0.0, requires_grad=True)
```

```python
z=w*x+b
z
```

```python
y_pred =torch.sigmoid(z)
y_pred
```

```python
# Binary Cross-Entropy Loss for scalar
def binary_cross_entropy_loss(prediction, target):
    epsilon = 1e-8  # To prevent log(0)
    prediction = torch.clamp(prediction, epsilon, 1 - epsilon)
    return -(target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
```

```python
loss=binary_cross_entropy_loss(y_pred,y)
loss
```

```python
loss.backward()
print(w.grad)
print(b.grad)
```
