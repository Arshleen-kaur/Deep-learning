```python
# Datasets and Dataloaders
# The issue with the code previously written is 
# -> We are using batch gradient descent; which is memory inefficient
# -> Not good convergence
# Solution: using batches of data to train the model
#  We can implement this manually; but there are issues with that code
# -> No easy way to apply transformations
# -> No standard interface of data
# -> Shuffling and sampling
# -> Batch management and parallelisation

# Datasets and DataLoader and core abstractions in PyTorch that decouple how you define your data and how you efficiently iterate over it in training loops
# __init__() which tells how data should be loaded
# __len__() which returns the total number of samples
# __getitem__(index) which returns the data and label at the given index

# DataLoader
# It divides indices into chunks of batch_size
```

```python
from sklearn.datasets import make_classification
import torch
```

```python
X,y=make_classification(
    n_samples=10,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_classes=2,
    random_state=42
)
```

```python
X =torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y,dtype=torch.long)
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):

    def __init__(self, features, labels):
        self.features = features
        self.labels=labels

    def __len__(self):
        return self.features.shape[0]
    def __getitem__(self, index):
        return self.features[index], self.labels[index]
```

```python
dataset = CustomDataset(X,y)
```
