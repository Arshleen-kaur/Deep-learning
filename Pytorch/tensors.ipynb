```python
# Tensors: Tensor is a data structure multi dimensional array designed for mathematical and computational efficiency
# O dimensional tensors: Scalar 1,2,3,4,5
# Vectors: 1 dimensional tensor (a list of numbers)
# Matrices: 2-dimensional tensors (grayscale image)
# 3D tensors: coloured images (rgb)
# 4D tensor: Batches of RGB image
# 5D tensor: Video data
# Why are tensors useful?
# -> Mathematical Operations
# -> Representation of real-world data
# -> Efficient Computations
# Where are Tensors used in Deep Learning?
# -> Data Storage
# -> Weights and Biases
# -> Matrix Operations
# -> Training Process
```
