```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_circles
```

```python
np.random.seed(42)
X,y= make_circles(n_samples=500, factor =0.1, noise=0.35, random_state=42)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
X.shape
```

```python
X[:5]
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtree=DecisionTreeClassifier(random_state=42)
dtree.fit(X_train,y_train)

plt.figure(figsize=(12,8))
x_range=np.linspace(X.min(),X.max(),100)
xx1,xx2=np.meshgrid(x_range,x_range)
y_hat=dtree.predict(np.c_[xx1.ravel(),xx2.ravel()]).reshape(xx1.shape)
plt.contourf(xx1,xx2,y_hat,alpha=0.3)
plt.scatter(X[:,0],X[:,1],c=y)
plt.title("Decision Tree Classifier Decision Boundary")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(n_estimators=100,random_state=42)
rf.fit(X_train,y_train)

plt.figure(figsize=(12,8))
X_range=np.linspace(X.min(),X.max(),100)
xx1,xx2=np.meshgrid(X_range,X_range)
y_hat=rf.predict(np.c_[xx1.ravel(),xx2.ravel()]).reshape(xx1.shape)
plt.contourf(xx1,xx2,y_hat,alpha=0.3)
plt.scatter(X[:,0],X[:,1],c=y)
plt.title("Random Forest Classifier Decision Boundary")
plt.show()
```
